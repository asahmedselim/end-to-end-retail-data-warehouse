STEP 3: Data Cleaning

```python
# Date Fixing
sales['Date'] = pd.to_datetime(sales['Date'], dayfirst=True)
features['Date'] = pd.to_datetime(features['Date'], dayfirst=True)
```

```python
# Delete Negative values in Weekly_sales
sales = sales[sales['Weekly_Sales'] >= 0]
```

```python
# Negative values in Markdown 1,2,3,5
markdown_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown5']

for col in markdown_cols:
    features[col] = features[col].apply(lambda x: 0 if x < 0 else x)
```

```python
# Missing values in Markdawns, CPI, Unemployment
markdown_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown4','MarkDown5']
features[markdown_cols] = features[markdown_cols].fillna(0)

features['CPI'] = features['CPI'].fillna(features['CPI'].mean())
features['Unemployment'] = features['Unemployment'].fillna(features['Unemployment'].mean())
```

```python
# Sorting
sales = sales.sort_values(by='Date')
features = features.sort_values(by='Date')
```
