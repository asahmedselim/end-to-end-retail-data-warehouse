# Transform Data to SQL Server

```python
# 1. Imports libraries
import pandas as pd
from sqlalchemy import create_engine
```

```python
# 2. Load Data
sales = pd.read_csv("sales_cleaned.csv")
features = pd.read_csv("features_cleaned.csv")
stores = pd.read_csv("stores_cleaned.csv")
```

```python
sales.info()
```

```python
features.info()
```

```python
# 3. Data Type Fix
sales['Date'] = pd.to_datetime(sales['Date'])
```

```python
features['Date'] = pd.to_datetime(features['Date'])
```

```python
sales.info()
```

```python
features.info()
```

```python
# 4. DB Connection
server = 'localhost\MSSQLSERVER06'
database = 'Walmart_DB'

engine = create_engine(
    f"mssql+pyodbc://{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server"
)
```

```python
# 5. Upload Tables
sales.to_sql('sales', engine, if_exists='replace', index=False)
```

```python
features.to_sql('features', engine, if_exists='replace', index=False)
```

```python
stores.to_sql('stores', engine, if_exists='replace', index=False)
```
